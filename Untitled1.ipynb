Ячейка 2: Основной код и тесты

```python
%%file test_homework.py
import pytest
import sqlite3

def get_text_length(text: str) -> int:
    """Возвращает количество символов в строке."""
    return len(text)

def write_data_to_file(filepath: str, content: str) -> None:
    """Записывает переданную строку в указанный файл."""
    with open(filepath, 'w', encoding='utf-8') as f:
        f.write(content)

@pytest.fixture
def db_connection():
    """
    Фикстура для создания подключения к in-memory базе данных SQLite.
    База в оперативной памяти (':memory:') создается заново для каждого теста,
    что гарантирует полную очистку данных между тестами.
    """
    conn = sqlite3.connect(':memory:')
    cursor = conn.cursor()
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS students (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT
        )
    ''')
    conn.commit()

    yield conn


    conn.close()

@pytest.mark.parametrize("test_input, expected_length", [
    ("", 0),
    ("     ", 5),
    ("Первая строка\nВторая строка", 27)
])
def test_get_text_length(test_input, expected_length):
    assert get_text_length(test_input) == expected_length


def test_write_data_to_file(tmp_path):
    temp_file = tmp_path / "my_test_file.txt"
    test_string = "Проверка записи в файл"

    write_data_to_file(str(temp_file), test_string)

    assert temp_file.read_text(encoding='utf-8') == test_string


def test_database_insert(db_connection):
    """Проверяем, что можем добавить запись в БД."""
    cursor = db_connection.cursor()
    cursor.execute("INSERT INTO students (name) VALUES ('Maksim')")
    db_connection.commit()

    cursor.execute("SELECT * FROM students")
    records = cursor.fetchall()

    assert len(records) == 1
    assert records[0][1] == 'Maksim'


def test_database_is_clean(db_connection):
    """
    Проверяем, что БД очистилась после предыдущего теста.
    Так как фикстура запускается заново, таблица должна быть пустой.
    """
    cursor = db_connection.cursor()
    cursor.execute("SELECT * FROM students")
    records = cursor.fetchall()

    assert len(records) == 0
```
